```python
# Unit 2(a) Part 1: Solving a definite integral by Gauss Legendre quadrature method.
import numpy as np

def integrand(t):
    return t**3 + 2*t

pts = 4
low, high = 0, 2

nodes, weights = np.polynomial.legendre.leggauss(pts)
t_mapped = 0.5 * (high - low) * nodes + 0.5 * (high + low)
ans = 0.5 * (high - low) * np.sum(weights * integrand(t_mapped))

print("Definite Integral =", round(ans, 4))
```

```python
# Unit 2(a) Part 2: Application - representation of a function as a linear combination of Legendre polynomials.
import numpy as np
from numpy.polynomial.legendre import leggauss, legval

def func(t):
    return 2*t**2 - t + 3

nodes, weights = leggauss(6)
coeffs = []

for deg in range(3):
    P = legval(nodes, [0]*deg + [1])
    c = ((2 * deg + 1) / 2) * np.sum(weights * func(nodes) * P)
    coeffs.append(round(c, 3))

print("f(x) =", coeffs[0], "P0 +", coeffs[1], "P1 +", coeffs[2], "P2")
```

```python
# Unit 2(b) Part 1: Solving improper integrals over the positive real axis using Gauss Laguerre quadrature method.
import numpy as np

def func(t):
    return t**3

nodes, weights = np.polynomial.laguerre.laggauss(5)
ans = np.sum(weights * func(nodes))

print("Gauss-Laguerre Integral =", round(ans, 4))
```

```python
# Unit 2(b) Part 2: Solving improper integrals over entire real axis using Gauss Hermite quadrature method.
import numpy as np

def func(t):
    return t**4

nodes, weights = np.polynomial.hermite.hermgauss(5)
ans = np.sum(weights * func(nodes))

print("Gauss-Hermite Integral =", round(ans, 4))
```
