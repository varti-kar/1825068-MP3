```python
# Unit 1(a): Determine the nth roots of a complex number and represent it in Cartesian and polar form.
import cmath
import math
import matplotlib.pyplot as plt

z_num = 16 + 0j
order = 4

r, phi = cmath.polar(z_num)
mag = r ** (1 / order)

xs, ys = [], []
for m in range(order):
    ang = (phi + 2 * math.pi * m) / order
    root = cmath.rect(mag, ang)
    xs.append(root.real)
    ys.append(root.imag)
    print(f"m={m} | Polar: ({round(mag, 2)}, {round(ang, 2)}) | Cartesian: {round(root.real, 2)} + {round(root.imag, 2)}j")

plt.plot(xs, ys, "ro")
plt.axis("equal")
plt.show()
```

```python
# Unit 1(b): 2D rotation of complex numbers
import matplotlib.pyplot as plt
import numpy as np

deg = 6
rad = np.radians(deg)
rot = np.array([[np.cos(rad), -np.sin(rad)], [np.sin(rad), np.cos(rad)]])

s = np.array([5.0, 3.0])
x, y = [s[0]], [s[1]]

for _ in range(60):
    s = rot @ s
    x.append(s[0])
    y.append(s[1])

# Minimalist plotting
plt.scatter(x, y, color="teal")
plt.grid(True)
plt.show()
```

```python
#Unit 1(c): plotting f(z) and w
import matplotlib.pyplot as plt
import numpy as np

z = np.array([2 + 1j, 3 + 2j, 4 + 1j, 5 + 3j, 3 + 4j])

w = z**3

plt.subplot(1, 2, 1)
plt.plot(z.real, z.imag,"ro-")
plt.title("z plane")

plt.subplot(1, 2, 2)
plt.plot(w.real, w.imag, "yo-")
plt.title("w plane (w = z^3)")

plt.show()
```
